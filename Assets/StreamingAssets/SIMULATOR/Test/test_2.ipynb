```python
import matplotlib.pyplot as plt

times = [1,2,3,4,5]
results = dict()

results[("USER_1",1)] = (10,1)
results[("USER_2",1)] = (11,2)
results[("USER_2",2)] = (21,6)
results[("USER_3",2)] = (19,7)
results[("USER_4",2)] = (21,8)
results[("USER_4",1)] = (10,4)
results[("USER_5",1)] = (11,5)
results[("USER_1",3)] = (10,1)
results[("USER_2",3)] = (11,2)
results[("USER_1",5)] = (10,1)
results[("USER_2",5)] = (11,2)
results[("USER_3",5)] = (9,3)
results[("USER_1",2)] = (20,5)
results[("USER_3",3)] = (9,3)
results[("USER_4",3)] = (10,4)
results[("USER_5",3)] = (11,5)
results[("USER_1",4)] = (20,5)
results[("USER_2",4)] = (21,6)
results[("USER_3",4)] = (19,7)
results[("USER_4",4)] = (21,8)
results[("USER_5",4)] = (20,9)
results[("USER_5",2)] = (20,9)
results[("USER_4",5)] = (10,4)
results[("USER_5",5)] = (11,5)
results[("USER_3",1)] = (9,3)



mean_capacities = []
mean_latencies = []
mean_times = []

for i in range(0, len(times)):

    capacities_for_mean, latencies_for_mean = [], []
    flag = False

    for current_key in results.keys():

        if times[i] == current_key[1]:

            capacities_for_mean.append(results[current_key][0])
            latencies_for_mean.append(results[current_key][1])

    mean_times.append(times[i])

    mean_c = sum(capacities_for_mean) / len(capacities_for_mean)
    mean_capacities.append(mean_c)

    mean_l = sum(latencies_for_mean) / len(latencies_for_mean)
    mean_latencies.append(mean_l)

plt.figure()
plt.plot(mean_times, mean_capacities, linewidth=2, label="Mean capacity (Mbps)")
plt.legend()

plt.figure()
plt.plot(mean_times, mean_latencies, linewidth=2, label="Mean prop latency (ms)")
plt.legend()
```
