```python
import networkx as nx

graph = nx.Graph()

graph.add_node("MACRO_BS_1")
graph.add_node("MACRO_BS_2")
graph.add_node("MACRO_BS_3")
graph.add_node("MACRO_BS_4")
graph.add_node("SMALL_BS_1")
graph.add_node("SMALL_BS_2")
graph.add_node("SMALL_BS_3")
graph.add_node("SMALL_BS_4")
#graph.add_edge("BS_1", "BS_2", name_edge="LINK_1")

#for BS in graph.nodes:
#    print(BS)

macro_nodes = []
for BS in graph.nodes:

    if BS.startswith("MACRO"):
        macro_nodes.append(BS)

for i in range(1, len(macro_nodes)):
    graph.add_edge(macro_nodes[i-1], macro_nodes[i])

print(graph.edges)
```
