```python
current_user_id = "USER_9"
connected_users = dict({"USER_1":1,"USER_2":2,"USER_3":2,"USER_4":2,"USER_5":5})
users_assigment_dict = dict({"USER_1":20,"USER_2":40,"USER_3":30,"USER_4":60,"USER_5":40})
available_bandwidth = 1
priority = 3
requirement = 190

aux_priority_dict = dict()

necessary_bandwidth = requirement - available_bandwidth
bw_assigned_to_user = 0

for USER_AUX in users_assigment_dict.keys():
    if priority > connected_users[USER_AUX]:
        aux_priority_dict[USER_AUX] = connected_users[USER_AUX]

for USER_AUX_2 in connected_users.keys():

    if USER_AUX_2 in aux_priority_dict:

        sum_inv_p_i = 0
        for PRIORITY in aux_priority_dict.values():
            sum_inv_p_i += 1/PRIORITY

        priority_factor = 1 / (aux_priority_dict[USER_AUX_2]*sum_inv_p_i)
        unallocated_bandwidth = necessary_bandwidth * priority_factor
        bw_assigned_to_user += unallocated_bandwidth

        if users_assigment_dict[USER_AUX_2] - unallocated_bandwidth > 0:
            users_assigment_dict[USER_AUX_2] = users_assigment_dict[USER_AUX_2] - unallocated_bandwidth

        else:
            bw_assigned_to_user += users_assigment_dict[USER_AUX_2]
            necessary_bandwidth = necessary_bandwidth - bw_assigned_to_user
            del aux_priority_dict[USER_AUX_2]
            users_assigment_dict[USER_AUX_2] = 0

users_assigment_dict[current_user_id] = bw_assigned_to_user + available_bandwidth

#print(users_assigment_dict)
#sum(connected_users.values())
```
